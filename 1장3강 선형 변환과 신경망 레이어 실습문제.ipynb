`xW`는 수식에서 x와 W의 행렬곱을 뜻하며, 코드에서는 `x @ W`로 씁니다. `xW+b`는 그 결과에 편향 b를 더하라는 뜻입니다. `T(x)`는 ‘입력 x에 T라는 계산을 적용한 결과’로 읽습니다.

````python
### ReLU ###

# 공부를 시작한 뒤 처음 2시간은 기초를 익히기 때문에 점수 상승을 기대하기 어렵다
# 기초를 충분히 파악한 2시간 뒤부터 점수가 올라갈 것이다.

# h = x - 2
# y = 50 + 10h
# y = 50 + 10(x - 2)
# y = 30 + 10x
# h = ReLU(x-2) = max(0, x-2)

# 공부시간 0시간 일 때
# x - 2 = -2, ReLU를 적용하면 0, 최종점수 = 50 --> 사실상 이건 계산 필요 없음(생략가능)

# 공부시간 1시간 일 때
# X -2 = -1, ReLU를 적용하면 0, 최종점수 = 50 --> 사실상 이건 계산 필요 없음(생략가능)

# 공부시간 2시간 일 때
# X-2 = 0, ReLU를 적용하면 0, 최종점수 = 50

# 공부시간이 3시간 일 때
# X-2 = 1, ReLU를 적용하면 1, 최종점수 = 60

# 공부시간이 4시간 일 때
# X-2 = 2, ReLU를 적용하면 2, 최종점수 = 70

# 2시간 이하의 공부시간이라면 굳이 패턴을 파악하지 않고 50점 유지
# 2시간 초과의 공부시간이라면 시간에 따른 점수 증가
````

````python
### 딥러닝 ### 

# 공부 기록으로 시험 점수를 예측하는 신경망 만들기
# 입력값은 3개(공부시간, 풀어본 문제 수, 복습 횟수)
# 값은 (2,3,1)이라고 가정
# 신경망에 사용되는 데이터 X1, X2, X3 = [2,3,1]

# 은닉층은 2개, 뉴런은 3개를 사용하여 딥러닝을 구성해보겠다
# 첫번째 은닉층, 뉴런 3개가 각각 계산하겠다
# 1. 세 뉴런 모두가 X의 특성값인 [2,3,1]을 받는다.
# 단, 각각의 뉴런은 가중치와 편향이 다르기 때문에 다른 결과를 얻는다.

# 첫번째 뉴런, 가중치는 [1,1,-1] 편향은 -1로 가정하면
# z1 = (1x2)+(1x3)+(-1x1)-1 = 3        <- -1은 편향임
# h1 = ReLU(3) = 3

# 두번째 뉴런, 가중치는 [-1, 1, 1] 편향은 0으로 가정
#  z2 = (-1 x 2) + (1 x 3) + (1 x 1) + 0 = 2
# h2 = ReLU(2) = 2

# 세번째 뉴런, 가중치는 [1, -1, -1], 편향은 1로 가정
# z3 = (1 x 2)+(-1 x 3)+(-1 x 1) + 1 = -1
# h3 = ReLU(-1) = 0
````

|뉴런|가중합+편향|ReLU 적용후|
|---|---|---|
|뉴런1|3|3|
|뉴런2|2|2|
|뉴런3|-1|0|

````python
# -> 첫번째 은닉층의 출력값은 [3,2,0]이다.
# -> 두번째 은닉층에서 사용할 중간값을 의미한다.

# 두번째 은닉층: 앞에서 나온 중간값 [3,2,0]을 받는다.
# 첫번째 뉴런: 첫번째 은닉층은 원래 데이터 [2,3,1]을 받았지만 두번째 은닉층에서는 첫번째 은닉층의 결과인 [3,2,0]을 받아서 계산했다.

# u1 = (1x3)+(1x2)+(1x0) - 1 = 4
# a1 = ReLU(4) = 4

# 두번째 뉴런: 
# u2 = (-1 x 3) + (1 x 2) + (1 x 0) + 0 = -1
# a2 = ReLU(-1) = 0

# 세번째 뉴런:
# u3 = (1 x 3) + (-1 x 2) + (1 x 0) + 1 = 2
# a3 = ReLU(2) = 2
````

|뉴런|가중합+편향|ReLU값 변화|
|---|---|---|
|뉴런1|4|4|
|뉴런2|-1|0|
|뉴런3|2|2|

````python
# 두 번째 은닉층의 출력은 [4,0,2]

# 두 개의 은닉층을 활용하여 최종 출력층을 만들어 본다면
# 출력층은 마지막(두번째 은닉층)에서 나온 [4, 0, 2]를 받는다.
# 출력층도 마찬가지로 뉴런의 가중치를 [10, -5, -5], 편향을 20이라고 가정해보면
# y = (10 x 4) + (-5 x 0) + (5 x 2) + 20 = 70점
# 최종 예측 점수는 70점으로 볼 수 있다.
````

````python
# 지금까지의 순서 정리
# 1. 입력층 -> 학생의 정보 전달 -> [2, 3, 1]
# 2. 첫번째 은닉층 -> 뉴런 3개가 각 가중합과 편향 계산 -> [3, 2, -1]
# 3. 첫번째 은닉층의 ReLU -> 음수 결과를 0으로 변경 -> [3, 2, 0]
# 4. 두번째 은닉층 -> 뉴런 3개가 각 가중합과 편향을 계산 -> [4, -1, 2]
# 5. 두번째 은닉층의 ReLU -> 음수 결과를 0으로 변경 -> [4, 0, 2]
# 6. 출력층 -> 마지막 가중합과 편향을 계산([4,0,2]를 더 정교하게 바꿈, 역전파 라고 함) -> 예측 점수 70점

# 출력층의 예측 점수와 실제 점수를 비교한 뒤 가중치와 편향을 수정한다.
# 이 학생의 실제 점수가 80점이었다면
# 예측값: 70점
# 실제정담: 80점
# 제곱오차: (70-80)² = 100
# -> 이 오차를 바탕으로 다시 각 가중치와 편향의 기울기 계산 
````

````python
# 아핀 변환:
# 선형 변환 결과(ax + b)에 일정한 벡터나 숫자 b(선형변한 결과가 예측과 다를 수 있으므로)를 더하는 계산 
#  --> 기울기 변화없이(기울기와 편향을 다시 구하는대신), 선자체를 벡터값만큼 더해서 정답에 가까운 쪽으로 옮기는 방법을 택함.
# 입력을 조합한 뒤에 기준 위치를 옮긴다.
# T(x) = xW + b
# 선형결과 [6,6]에 b = [1, -2]를 더하면 [7,4]
# 왜 쓰냐하면, 화면 속 점 [3,2]를 가로, 세로가 n배 늘어났고 옮긴다고 가정하에 아핀 변환을 사용할 수 있다.
# 
# 뉴런:
# 입력을 받아 출력 하나를 만드는 계산 단위
# ex) z = x1w1 + x2w2 + b <-- 이게 여러개 있는거라고 생각하면 됨
# 학생의 [출석, 과제] 기록 [3,2]를 받아 중간 값(은닉층 출력 결과)를 만드는 계산기가 있다.
# 가중치 [1,2]를 곱해 3+4 = 7을 만들고 편향 -4를 통해 3을 얻는다.
#
# layer와 신경망
# layer: 여러 뉴런을 모아 놓은 계산 단계
# 신경망: 이런 계산들을 연결한 모델
# 예: 학생의 기록 [3,2]를 단계별 계산을 하면
# 첫 단계의 뉴런 2개가 각각 3과 -3을 만들고, ReLU를 거쳐 3,0을 만들어 내면
# 두 번째 단계에서는 [3,0]을 이어받아서 3x2+0x(-1)+1 = 7을 만든다
# 학생 기록 -> 중간 숫자 -> 최종 점수 <--- 이 각 단계가 layer  

# 활성화 함수:
# 뉴런의 가중합에 편향을 더한 값을 다시 바꾸는 규칙
# ex) z = xw + b를 구하고 난 뒤 a = f(z)를 계산
# 학습실 이용 시간을 4시간 초과한 경우 금액을 부과하기 위해 따로 기록을 남긴다고 가정하면
# z = -4를 계산, 3시간을 이용했다면 z = -1로 추가적인 금액은 부과하지 않겠다.
# 이렇게 이미 계산된 식에 임의의 식을 더해서 그 기록과 규칙을 바꾸려고 하는 게 활성화 함수다.

# 비선형 함수:
# 선형 변환이 지켜야하는 덧셈 및 배수 관계를 모두 만족하지 않는 함수
# 추가 이용 시간 y = ReLU(이용시간 - 4)를 계산한다고 가정하면
# 3시간 이용자 수, 4시간 이용자 수는 추가 시간이 0인데
# 5시간, 6시간 이용자 수는 1시간, 2시간
# 이용 시간이 한 시간 늘어나도 기준시간은 결과가 그대로이기 때문에 4시간 전까지 올바른 선형관계를 보이기 어렵다.

# 은닉층(hidden layer):
# 입력과 최종 출력 사이에서 중간 값을 만드는 계산 단계
# 은닉이라는 단어가 들어가 있지만 print()로 중간값들을 확인할 수는 있음
# 학생의 출석 및 과제 기록이 [3,2]로 최종 점수를 바로 만들지 않고 중간 계산을 거쳐서 검토한다고 가정해보면
# 두 뉴런으로 [3, -3]값을 만든 뒤 ReLU를 적용하여 [3,0]을 만든다 <- 이 과정을 은닉층이라고 부른다(z = x1w1 + x2w2 + b 이거 여러개 값들의 벡터).
# 출력층(최종 산출)은 이 [3,0]을 조합하여 최종 점수를 만든다
# 실제 신경망 계산에서는 입력을 여러 중간 값으로 바꾸어 계산을 진행한다.

# 출력층:
# 모델의 최종 결과를 만드는 계산 단계
# 앞 단계에서 학생의 중간 값 H = [5 3]을 얻었다고 가정해보면 이 각 컬럼에 대한 점수들을 가지고 실제 점수에 반영해보겠다.
# 10점 만점이라고 가정하고 가중치를 [2, -1], 편향을 1로 가정하면
# 5x2 + 3x-1 + 1 = 8점, 최종적으로 학생의 점수는 8점으로 나타낼 수 있다.
# 출력층에 ReLU가 없다면(중간값에는 ReLU를 적용하고) 만약에 H = [0,3]이면 0-3+1 = -2로, 최종 출력값은 0이 아니라 -2로 계산될 수 있다.

# 순전파(Forward Propagation):
# 입력부터 최종 출력까지 계산을 앞 방향으로 진행하는 과정
# 학생 기록 [3,2]를 모델에 넣고 계산 결과를 순전파로 따라가보면
# 1. 가중합과 편향을 계산 z1 = [3, -3]
# 2. ReLU를 적용하여 H = [3,0]을 만들고
# 3. 출력층에서 3x2+0x(-1)+1 = 7을 산출한다
# 1-> 2 -> 3 이 과정이 순전파라고 부른다.
````

## 필수 1. 선형·아핀 변환과 파라미터 이해하기

### 문제 1-1. 선형 변환에 편향을 더해 보기

#### 문제 설명

교안 1절의 예제입니다. 선형 변환은 입력에 가중치 행렬을 곱하는 계산이며, 아핀 변환은 그 결과에 편향을 더하는 계산입니다.

```python
x = np.array([3, 2])
W = np.array([[2, 0], [0, 3]])
bias = np.array([1, -2])
zero = np.array([0, 0])
```

- 선형 변환: `x @ W`
- 아핀 변환: `x @ W + bias`

W의 첫 번째 열은 [2,0], 두 번째 열은 [0,3]입니다. 0벡터는 모든 값이 0인 벡터를 말합니다.

#### 요구사항

- 기본 점수, 두 변경의 점수와 변화량, 0벡터의 점수

````python
# 문제 1-2. 가중치와 편향을 하나씩 바꾸기
# 위의 데이터와 작성 도움말을 보고 한 단계씩 코드를 작성하세요.
# 중간 결과를 출력하여 확인하세요.

X = np.array([
    [4, 2],
    [1, 5]]) # <- 행마다 고객 한명 배정
w = np.array([2, -1]) # 가중치 배정 첫 입력에 2, 둘째 입력에 -1
b = 3 # 고객마다 곱셈 결과를 더한 뒤 3을 더함

base_scores = X @ w + b # 각 행의 두 입력과 w의 두 가중치를 짝지어 계산
print("기본 점수:", base_scores) 

w_new = np.array([2, -2]) # 가중치 배정 두번째 가중치만 -2로 수정
weight_scores = X @ w_new + b # 가중치 변화에 따른 변화량 산출
print("가중치 변경 점수:", weight_scores)
print("가중치 변경에 따른 변화:", weight_scores - base_scores) # 가중치 변화에 따른 변화량 산출 

b_new = 5 # 새로운 편향
bias_scores = X @ w + b_new # 기본 가중치로 돌아가 편향만 수정
print("편향 변경 점수:", bias_scores)
print("편향 변경에 따른 변화:", bias_scores - base_scores)

zero = np.array([0, 0]) # 0 백터 점수 확인
print("기본 설정의 0벡터 출력:", zero @ w + b)
````

### 문제 2-1. ReLU와 한 레이어를 순서대로 계산하기

교안 3절의 예제입니다. 먼저 ReLU의 규칙을 연습한 뒤 레이어 하나를 계산합니다.

**ReLU(렐루)의 규칙:** 양수는 그대로 두고, 음수는 0으로 바꿉니다. 0은 0으로 둡니다.

```python
def relu(z):
    return np.maximum(0, z)

test_z = np.array([-3, -1, 0, 2, 5])
X = np.array([[3, 2], [1, 4], [0, 2]])
W = np.array([[1, -2], [2, 1]])
b = np.array([-4, 1])
```

X의 세 행은 고객 A, B, C입니다. W의 두 열은 뉴런 두 개의 가중치입니다. 첫 뉴런의 가중치는 [1,2], 둘째 뉴런의 가중치는 [−2,1]입니다. 편향은 각각 −4와 1입니다.

이번 계산은 **가중치 적용 → 편향 덧셈 → ReLU**의 세 단계입니다.

5. `Z.shape`와 `A.shape`를 출력하고, 고객 A의 두 뉴런 계산을 손으로 확인하세요.

`def relu(z):`와 그 아래 `return` 줄은 위 예제를 그대로 사용하세요. return 앞의 들여쓰기도 유지하세요. 고객 A의 첫 뉴런은 `3×1 + 2×2`를 구한 뒤 −4를 더합니다. 마지막에 그 값과 0 중 큰 값을 고릅니다.

#### 해석 질문

**Q1.** ReLU는 [−3,−1,0,2,5]를 어떻게 바꾸나요?

**Q3.** 고객 B와 C의 Z와 ReLU 결과는 각각 무엇인가요?

**Q4.** Z와 A의 Shape (3,2)는 무엇을 뜻하나요? 음수를 0으로 바꾸면 배열의 크기도 줄어드나요?

#### 제출 결과

- ReLU 연습 결과, weighted·Z·A의 출력, Z·A의 Shape
- 고객 A의 계산 과정과 Q1~Q4 답변

````python
# 문제 2-1. ReLU와 한 레이어를 순서대로 계산하기
# 위의 데이터와 작성 도움말을 보고 한 단계씩 코드를 작성하세요.
# 중간 결과를 출력하여 확인하세요.

def relu(z):
    return np.maximum(0, z) # z와 0 값을 비교하여 큰 값을 Return

test_z = np.array([-3, -1, 0, 2, 5]) # 예로 보면 0, 0, 0, 2, 5
print("테스트 입력:", test_z)
print("테스트 ReLU:", relu(test_z))

X = np.array([
    [3, 2],
    [1, 4],
    [0, 2]]) # 고객 3명, 입력값 2개(피처가 2개)

W = np.array([
    [1, -2],
    [2,  1]]) # w의 첫열 [1,2] <- 첫 뉴런
              # w의 둘째 열 [-2, 1]은 둘째 뉴런에 사용
              # 두 뉴런은 같은 고객의 입력 2개를 각각 받아 서로 다른 가중치로 계산한다.

b = np.array([-4, 1]) # 
print("X shape:", X.shape) 
print("W shape:", W.shape)
print("b shape:", b.shape)

weighted = X @ W
Z = weighted + b # 편향이 적용되는 위치 확인  
A = relu(Z)
print("편향 전 행렬곱:")
print(weighted)
print("편향을 더한 Z:")
print(Z)
print("ReLU 결과 A:")
print(A)
print("Z shape:", Z.shape)
print("A shape:", A.shape)
````

- 최종 산출된 점수는 A, B, C 순서대로 7, 8, -2
